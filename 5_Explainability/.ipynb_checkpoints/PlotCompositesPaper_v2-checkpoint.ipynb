```python
def plot_composite_in_ax(ax_temp,ivar,week_out,wr,season,letter_title,shrink=0.6,cbar=False,cbar_ax=None):
    path_nc_anoms = f'{path_weekly_anoms}{list_vars[ivar]}.nc'
    anoms = xr.open_dataset(path_nc_anoms)
    anoms = anoms.assign_coords(time=pd.DatetimeIndex(anoms.time).normalize())
    var_name_nc = list(anoms.data_vars.keys())[0]
    week_out_str = f'week{week_out}'
    shap_full = xr.open_dataset(f'{path_shap}/SHAP_{list_vars[ivar]}_{week_out_str}.nc')
    
    df_results_full = pd.read_csv(f'{path_results}/Results_{list_vars[ivar]}_{week_out_str}.csv',
                                 index_col=0, parse_dates=True)
    df_results_temp = copy.deepcopy(df_results_full)
    if season!='All':
        df_results_temp = df_results_temp.loc[(df_results_temp.index.month==indices_seasons[season][0])|\
                        (df_results_temp.index.month==indices_seasons[season][1])|\
                        (df_results_temp.index.month==indices_seasons[season][2])]
    anoms_temp = anoms.sel(time=df_results_temp.index)
    where_correct = np.where((df_results_temp['y_predicted']==df_results_temp['y_true'])&
                            (df_results_temp['y_predicted']==wr))[0]
    if len(where_correct)==0:
        composite_plot = copy.deepcopy(anoms[var_name_nc].isel(time=0))*np.nan
    else:
        composite_plot = anoms_temp[var_name_nc]
        composite_plot = composite_plot.isel(time=where_correct)
        composite_plot = composite_plot.mean('time')#.fillna(0)
        symmetric_max = np.max(np.abs(np.max([abs(np.nanquantile(composite_plot,0.001)),\
                                              abs(np.nanquantile(composite_plot,0.999))])))
    compositevals = composite_plot.values
    compositevals[:,-1] = compositevals[:,-2]
    compositevals[0,:] = compositevals[1,:]
    compositevals[-1,:] = compositevals[-2,:]
    
    ##### Plot ####
    intervals = 41
    mini=-1
    maxi=1
    bounds=np.linspace(mini,maxi,intervals)
    # print(vmax,vmin)
    cs = ax_temp.contourf(composite_plot.lon,composite_plot.lat,\
        compositevals,cmap='seismic',
        transform=ccrs.PlateCarree(),
        vmin=mini,vmax=maxi,extend='both',levels=bounds)
    ax_temp.coastlines(resolution='110m', color='k', linewidth=0.75, zorder=10)
    ax_temp.margins(x=0, y=0)

    # Add colorbar
    # Add colorbar
    if cbar==True:
        # Add colorbar
        cbar_ax = fig.add_axes(cbar_ax)
        ticks_1 = [-1,0,1]
        cbar = fig.colorbar(cs, cax=cbar_ax, ticks=ticks_1[:],
                        orientation='vertical', extend='both')
        cbar.set_label("Anomaly [$\sigma$]")
    
    # Set geographical extent (lon_min, lon_max, lat_min, lat_max)
    ax_temp.set_extent([100, -30, -28, 90], crs=ccrs.PlateCarree())

    name_var = list_vars[ivar].replace('_ERA5','').replace('_SODA','')
    name_var = name_var.replace('SWVL','SWC').replace('STL','ST')
    ax_temp.set_title(f'{letter_title}) {name_var} Anoms. - {season} - Week {week_out}\n{wr_names[wr]} - # Correct cases: {len(where_correct)}',loc='left')    
```

```python
def coarsen_geospatial_dataset(dataset, lat_factor, lon_factor,var_name_nc, aggregation_func=np.mean):
    """
    Coarsens a geospatial xarray Dataset with longitude wrapping, avoiding edge effects.

    Args:
        dataset (xr.Dataset): The input Dataset with 'lat' and 'lon' coordinates.
        lat_factor (int): The coarsening factor for the latitude dimension.
        lon_factor (int): The coarsening factor for the longitude dimension.
        aggregation_func (function): The aggregation function (e.g., np.mean, np.sum).

    Returns:
        xr.Dataset: The coarsened Dataset.
    """
    dataset = dataset.isel(lon=slice(1,-2))
    # Extend the data horizontally
    lon_extension = 30  # Extend by half the longitude coarsening factor on each side
    # return lon_extension
    original_lon = dataset['lon'].values
    lon_range = original_lon.max() - original_lon.min()
    # return lon_range
    # Create artificial longitudes beyond the original range.
    extended_lon_left = original_lon[:lon_extension] - 360
    extended_lon_right = original_lon[-lon_extension:] + 360

    extended_lon = np.concatenate([extended_lon_left, original_lon, extended_lon_right])
    # plt.plot(extended_lon)
    extended_data = xr.concat([dataset[var_name_nc].isel(lon = slice(-lon_extension, None)), dataset[var_name_nc],
                               dataset[var_name_nc].isel(lon = slice(0, lon_extension))], dim = 'lon')
    extended_data['lon'] = extended_lon
    # return extended_data
    extended_dataset = xr.Dataset({var_name_nc: extended_data}, coords={'lat': extended_data['lat'], 'lon': extended_lon, 'time': extended_data['time']})
    # return extended_dataset
    # # Coarsen the extended dataset
    coarsened_extended = extended_dataset.coarsen(lat=lat_factor, lon=lon_factor, boundary='pad').reduce(aggregation_func)

    # # Trim the extended coarsened data back to the original longitude range
    coarsened_dataset = coarsened_extended.sel(lon=slice(original_lon.min(), original_lon.max()))

    return coarsened_dataset
```

```python
def get_weighted_area_average(da):
    
    # Compute the cosine of the latitudes (in radians) for weighting
    # For a rectangular grid the cosine of the latitude is proportional to the grid cell area
    weights = np.cos(np.deg2rad(da.lat))
    weights.name = "weights"

    # Apply the weights and compute the mean across lat/lon
    da_weighted = da.weighted(weights)
    weighted_mean = da_weighted.mean(dim=['lat', 'lon'])
    
    return weighted_mean

def get_weighted_area_std(da):
    
    # Compute the cosine of the latitudes (in radians) for weighting
    # For a rectangular grid the cosine of the latitude is proportional to the grid cell area
    weights = np.cos(np.deg2rad(da.lat))
    weights.name = "weights"

    # Apply the weights and compute the mean across lat/lon
    da_weighted = da.weighted(weights)
    weighted_std = da_weighted.std(dim=['lat', 'lon'])
    
    return weighted_std
```

```python
def plot_shap_in_ax(ax_temp,ivar,week_out,wr,season,letter_title,shrink=0.6,cbar=False,cbar_ax=None):
    path_nc_anoms = f'{path_weekly_anoms}{list_vars[ivar]}.nc'
    anoms = xr.open_dataset(path_nc_anoms)
    anoms = anoms.assign_coords(time=pd.DatetimeIndex(anoms.time).normalize())
    var_name_nc = list(anoms.data_vars.keys())[0]
    week_out_str = f'week{week_out}'
    shap_full = xr.open_dataset(f'{path_shap}/SHAP_{list_vars[ivar]}_{week_out_str}.nc')
    
    var_name_nc = list(shap_full.data_vars.keys())[0]
    df_results_full = pd.read_csv(f'{path_results}/Results_{list_vars[ivar]}_{week_out_str}.csv',
                                 index_col=0, parse_dates=True)
    df_results_temp = copy.deepcopy(df_results_full)
    if season!='All':
        where_temp = np.where((df_results_temp.index.month==indices_seasons[season][0])|\
                        (df_results_temp.index.month==indices_seasons[season][1])|\
                        (df_results_temp.index.month==indices_seasons[season][2]))[0]
        df_results_temp = df_results_temp.loc[(df_results_temp.index.month==indices_seasons[season][0])|\
                        (df_results_temp.index.month==indices_seasons[season][1])|\
                        (df_results_temp.index.month==indices_seasons[season][2])]
        shap_temp = shap_full.isel(time=where_temp)
    else:
        shap_temp = copy.deepcopy(shap_full)
    
    shap_temp = coarsen_geospatial_dataset(shap_temp,3,3,var_name_nc,np.sum)
    shap_temp_std = get_weighted_area_std(shap_temp[var_name_nc])
    shap_temp[var_name_nc] = (shap_temp[var_name_nc])/shap_temp_std
    where_correct = np.where((df_results_temp['y_predicted']==df_results_temp['y_true'])&
                            (df_results_temp['y_predicted']==wr))[0]
    if len(where_correct)==0:
        shap_plot = copy.deepcopy(shap_full[var_name_nc].isel(time=0))*np.nan
    else:
        shap_plot = shap_temp[var_name_nc]
        shap_plot = shap_plot.isel(time=where_correct)
        shap_plot = shap_plot.mean('time').fillna(0)
        
        arr_temp = copy.deepcopy(shap_plot.values)
        arr_temp = gaussian_filter(arr_temp, sigma=0.75)
        symmetric_max = 1
        
    arr_temp = arr_temp
    arr_temp[:,0] = arr_temp[:,1]
    arr_temp[:,-1] = arr_temp[:,-2]
    arr_temp[0,:] = arr_temp[1,:]
    arr_temp[-1,:] = arr_temp[-2,:]
    data_cyclic, lon_cyclic = add_cyclic_point(arr_temp, coord=shap_plot.lon.values)
    
    intervals = 41
    mini=-symmetric_max
    maxi=symmetric_max
    bounds=np.linspace(mini,maxi,intervals)
    cs = ax_temp.pcolormesh(lon_cyclic,shap_plot.lat,\
        data_cyclic,cmap='bwr',
        transform=ccrs.PlateCarree(),
        vmin=mini,vmax=maxi)
    ax_temp.coastlines(resolution='110m', color='k', linewidth=0.75, zorder=10)
    ax_temp.margins(x=0, y=0)

    
    # Add colorbar
    if cbar==True:
        # Add colorbar
        cbar_ax = fig.add_axes(cbar_ax)
        ticks_1 = [-1,0,1]
        cbar = fig.colorbar(cs, cax=cbar_ax, ticks=ticks_1[:],
                        orientation='vertical', extend='both')
        cbar.set_label("$\Delta$P [$\sigma$]")
    
    # Set geographical extent (lon_min, lon_max, lat_min, lat_max)
    ax_temp.set_extent([100, -30, -28, 90], crs=ccrs.PlateCarree())
    name_var = list_vars[ivar].replace('_ERA5','').replace('_SODA','')
    name_var = name_var.replace('SWVL','SWC').replace('STL','ST')
    ax_temp.set_title(f'{letter_title}) {name_var} SHAP - {season} - Week {week_out}\n{wr_names[wr]} - # Correct cases: {len(where_correct)}',loc='left')
    return shap_temp
```
